# Applied Regularization CNN (Task 5)

### Import Libraries

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import os

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

from tensorflow.keras.callbacks import EarlyStopping

import numpy as np
from tensorflow.keras.utils import load_img, img_to_array
```

### Prepare Dataset

1. Load Dataset

```python
# Train
print("Train Dataset:")
train_dataset = tf.keras.utils.image_dataset_from_directory(
    "seg_train/seg_train",
    image_size=(150, 150),
    validation_split=0.2,
    subset="training",
    seed=42
)

# Validate
print("\nValidate Dataset:")
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "seg_train/seg_train",
    image_size=(150,150),
    batch_size=32,
    validation_split=0.2,
    subset="validation",
    seed=42
)

print("\nTest Dataset:")
# Test
test_dataset = tf.keras.utils.image_dataset_from_directory(
    "seg_test/seg_test",
    image_size=(150, 150),
    batch_size=32,
    shuffle=False
)

class_names = train_dataset.class_names
num_classes = len(class_names)
```

2. Normalization (Using Normalization from task 4)

Added Data Augmentation:
- **RandomFlip:** Helps the model recognize an object even when its position/orientation is reversed.
- **RandomRotation:** Makes the model less sensitive to small changes in the camera angle.
- **RandomZoom:** Helps the model recognize objects whether they appear closer or farther away.
- **RandomBrightness:** Helps the model handle images taken under different lighting conditions.

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomBrightness(0.1)
])

normalization_layer = tf.keras.layers.Rescaling(1./255)

def augment_and_normalize(x, y):
    x = data_augmentation(x, training=True)
    x = normalization_layer(x)
    return x, y

def normalize_data(x, y):
    x = normalization_layer(x)
    return x, y



train_dataset = train_dataset.map(augment_and_normalize)

validation_dataset = validation_dataset.map(normalize_data)

test_dataset = test_dataset.map(normalize_data)
```

3. Example Images

```python
plt.figure(figsize=(12, 8))

for images, labels in train_dataset.take(1):
    for i in range(12):
        ax = plt.subplot(3, 4, i + 1)

        plt.imshow(images[i])
        plt.title(class_names[labels[i]])
        plt.axis("off")

plt.show()
```

### CNN model

1. Build a model (Using the modified model from task 3)

Added Dropout rate

```python
model = Sequential([

    Conv2D(32, (3, 3), activation="relu", input_shape=(150, 150, 3)),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation="relu"),

    # Added Dropout rate
    Dropout(0.3),

    Dense(num_classes, activation="softmax")
    
])
```

2. Compile and Train
- Use **Sparse Categorical Cross Entropy** if the labels look like 3,6,2,...
- Use **Categorical Cross Entropy** if the labels look like [0,1,0,...] [0,...,1,0]

Added Early Stopping

```python
# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Added Early Stopping
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

# Train
history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=20,
    callbacks=[early_stopping]
)
```

3. Evaluate

```python
test_loss, test_accuracy = model.evaluate(test_dataset)

print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```

4. Save

```python
model.save("CNN_Regularized.keras")
```
