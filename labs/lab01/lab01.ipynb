```python
import matplotlib.pyplot as plt
import pandas as pd



df = pd.read_csv('lab01_dataset.csv')

print("First 5 rows")
print(df.head())

print("\n Basic Info")
print(df.info())
print(df.describe())

plt.figure(figsize=(6, 4))
df['Topic'].value_counts().plot(kind='barh', color='#4C72B0')

plt.title('Topic Count')
plt.xlabel('Count')
plt.ylabel('Topic')

plt.show()
```
